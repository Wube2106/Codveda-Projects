# Task 3: Neural Network using TensorFlow/Keras

## Objective
To build a feed-forward neural network using TensorFlow/Keras to classify handwritten digits (0–9) from the MNIST dataset. The model will be trained using backpropagation and evaluated using accuracy and loss visualization.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Flatten
```

## Step 2: Load MNIST Dataset

The MNIST dataset contains handwritten digits (0–9).
It is split into training and testing sets automatically.

```python
#load the dataset
(X_train,y_train),(X_test,y_test)=mnist.load_data()

```

## Step 3: Data Preprocessing

We normalize pixel values from 0–255 to 0–1 to improve training performance and stability.

```python
X_train=X_train/255.0
X_test=X_test/255.0
```

## Step 4: Build Neural Network Model

We create a feed-forward neural network with:
- Flatten layer (input layer)
- Two hidden layers with ReLU activation
- Output layer with Softmax for classification

```python
#build the neural network
model=Sequential([  Flatten(input_shape=(28,28)),
  Dense(128,activation='relu'),
  Dense(64,activation='relu'),
  Dense(10,activation='softmax')
])
```

## Step 5: Model Summary

This shows the architecture of the neural network including layers and number of parameters.

```python
model.summary()
```

## Step 6: Compile Model

We define:
- Optimizer: Adam (efficient gradient descent)
- Loss function: Sparse categorical crossentropy (for multi-class classification)
- Metric: Accuracy

```python
#compile the model
model.compile(
  optimizer='adam',
  loss='sparse_categorical_crossentropy',
  metrics=['accuracy']
)
```

## Step 7: Train the Model

The model learns using backpropagation.
We also use validation split to monitor performance on unseen data.

```python
#train the model
history=model.fit(
  X_train,
  y_train,
  epochs=10,
  validation_split=0.2
)
```

## Step 8: Evaluate Model

We test the model on unseen test data and measure accuracy.

```python
# Evaluate the model on test data
test_loss, test_accuracy=model.evaluate(X_test,y_test)
print("Test Accuracy:",test_accuracy)
```

## Step 9: Make Predictions

The model predicts probabilities for each digit.
We convert them into actual class labels using argmax.

```python
# make prediction
pred=model.predict(X_test)
predicted_labels=np.argmax(pred,axis=1)

print("Predicted:",predicted_labels[:10])
print("Actual  :",y_test[:10])
```

## Step 10: Visualize Training Performance

We plot:
- Training vs Validation Loss
- Training vs Validation Accuracy

```python
# Plot training vs validation loss and accuracy
plt.figure(figsize=(12,5))

# Loss
plt.subplot(1,2,1)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Loss Curve')
plt.legend()

# Accuracy
plt.subplot(1,2,2)
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Accuracy Curve')
plt.legend()

plt.show()


```

## Step 11: Confusion Matrix

The confusion matrix shows how well the model distinguishes between different digits.

```python
# Confusion Matrix
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cm = confusion_matrix(y_test, predicted_labels)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)

disp.plot(cmap='Blues')
plt.title("Confusion Matrix")
plt.show()
```
