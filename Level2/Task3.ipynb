# Task 3:K-Means Clustering

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
```

## Load the dataset

```python
# load the dataset
df = pd.read_csv(r"C:/Users/Hp/Documents/My project/Codveda Projects/Dataset/1) iris.csv")
df.head()
df.info()
```

## Handling the dataset
I checked the missing values and duplicates and then try to solve them.After that, since the algorithm is K-Means(unsuperivised ml), I remove the species columns since it is not useful.

```python
# handling the dataset
print("Before handling missing values:",df.isnull().sum())
print("Before handling duplicates:",df.duplicated().sum())
df=df.drop_duplicates()
df=df.drop(columns="species",axis=1,errors="ignore")
X=df
print("After handling duplicates:",df.duplicated().sum())
```

## Feature Scaling

```python
# Feature scaling
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
print(X.head())
print(X_scaled[0])
```

## Train the model using K -Means
I apply K-Means clustering and determine the optimal
number of clusters using the elbow method.

```python
#Train the model with K-Means and elbow method
wcss=[]
for k in range(1,11):
  kmeans=KMeans(n_clusters=k,random_state=42,init="k-means++")
  kmeans.fit(X_scaled)
  
  wcss.append(kmeans.inertia_)

```

## Elbow Graph
To determine the optimal number of clusters, I use the elbow graph.

```python
# visualization of the elbowm method
plt.figure(figsize=(8,5))
plt.plot(range(1,11),wcss,marker="o")
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.show()
```

```python
# choose k from elbow graph
kmeans=KMeans(n_clusters=2,random_state=42,init="k-means++")
clusters=kmeans.fit_predict(X_scaled)

df["Cluster"]=clusters
print(df.head())
```

## Reduce Dimension

```python
#Reduce Dimensions for Visualization
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
```

## Visualize Clusters
I use 2D scatter plot to visualize the the clusters.

```python
#Visualize Clusters


plt.figure(figsize=(10,6))

sns.scatterplot(
    x=X_pca[:,0],
    y=X_pca[:,1],
    hue=clusters,
    palette='Set1',
    s=100
)
```

## Plot Centroids

```python
# Plot centroids
centroids = pca.transform(
    kmeans.cluster_centers_
)

plt.scatter(
    centroids[:,0],
    centroids[:,1],
    marker='X',
    s=300,
    label='Centroids'
)

plt.title("K-Means Clusters")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.legend()
plt.show()
```

## Interpretation of Clusters

```python
#interepretation of clusters
print("Cluster Interpretation \n",df.groupby(by="Cluster").mean())
```
