# Task 3:KNN Classifier for iris data

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix
```

# Load the dataset

```python
# load the dataset
df = pd.read_csv(r"C:/Users/Hp/Documents/My project/Codveda Projects/Dataset/1) iris.csv")
df.head()
df.info()
```

# Handling dataset
I check both missing values and duplicates.

```python
# handling the dataset
print("Before handling:",df.isnull().sum())
print("Before Duplicates",df.duplicated().sum())
df=df.drop_duplicates()
print("After Duplicates",df.duplicated().sum())
```

# Feature selection

```python
# Feature selection 
X=df.drop("species",axis=1)
y=df["species"]
```

# Train-Test split and Feature scaling
I split the dataset into train and test and I scale features usign standard score(Z-score).

```python
# Train test split and Feature scaling
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,shuffle=True,random_state=42)
scalar=StandardScaler()

X_train=scalar.fit_transform(X_train)
X_test=scalar.transform(X_test)
```

# Train using KNN and Evaluate
I use the K- Nearest Neighbors(KNN) Classifier algorithm to train the model and I tried to train for different values of K.I evaluate the model using different metrics.

```python
# Try different K values
k_values = [1,3, 5, 7,15]
results=[]
for k in k_values:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='macro')
    recall = recall_score(y_test, y_pred, average='macro')
    cm = confusion_matrix(y_test, y_pred)

    print(f" For K= {k}")
    print(f"Accuracy: {acc:.3f}")
    print(f"Precision (macro): {precision:.3f}")
    print(f"Recall (macro): {recall:.3f}")
    print("Confusion Matrix:\n", cm)
    results.append([k,acc,precision,recall])
print(results)
```

# Compare results

```python
results_df=pd.DataFrame(results,columns=["k","acc","precision","recall"])
print("Comparsion Table")
print(results_df)
```
