KNN Classifier

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix
```

# Load the dataset

```python
# load the dataset
df = pd.read_csv(r"C:/Users/Hp/Documents/My project/Codveda Projects/Dataset/1) iris.csv")
df.head()
df.info()
```

```python
# handling the dataset
print("Before handling:",df.isnull().sum())
print("Before Duplicates",df.duplicated().sum())
df=df.drop_duplicates()
print("After Duplicates",df.duplicated().sum())
```

```python
# Feature selection 
X=df.drop("species",axis=1)
y=df["species"]
```

```python
# Train test split and Feature scaling
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,shuffle=True,random_state=42)
scalar=StandardScaler()

X_train=scalar.fit_transform(X_train)
X_test=scalar.transform(X_test)
```

```python
# Train the model with KNN
model=KNeighborsClassifier(n_neighbors=5)
model.fit(X_train,y_train)

y_pred=model.predict(X_test)
print(y_test[:10])
print(y_pred[:10])
```

```python
# Evaluation
accuracy=accuracy_score(y_test,y_pred)
precision=precision_score(y_test,y_pred,average="macro")
recall=recall_score(y_test,y_pred,average="macro")
cm=confusion_matrix(y_test,y_pred)

print("Accuracy:",accuracy)
print("Precision:",precision)
print("Recall:",recall)
print("Confusion Matrix:",cm)
```
