# Task 2:Build a Simple Regression Model

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Load the Dataset 
loading the dataset and check the information, columns

```python
#load the dataset
df=pd.read_csv("C:/Users/Hp/Documents/My project/Codveda Projects/Dataset/2) Stock Prices Data set.csv")
df.head()
df.describe()
df.info()
```

# Data Handling
We use the median value to handle missing values since the data is highly skewed.

```python
#handling missing values and duplicates
#handle missing values
print("Missing values Before:",df.isnull().sum())
numerical_features=df.select_dtypes(include=["number"]).columns.tolist()
for col in numerical_features:
  #we use median since the features are highly skewed
  if df[col].isnull().sum()>0:
    df[col].fillna(df[col].median(),inplace=True)  

print("Missing values After:",df.isnull().sum())

#handle duplicate values
print("Duplicates before:",df.duplicated().sum())
df=df.drop_duplicates()
print("Duplicates after",df.duplicated().sum())


```

# Input features and Target
The symbol is excluded because it has high cardinality and encoding will increase dimenstinality

```python
#input features and target
#The symbol is excluded because it has high cardinality and encoding will increase dimentinality
X=df.drop(["symbol","date","close"],axis=1)
y=df["close"]
print(X.shape)
print(y.shape)
```

# Train and Test Split
We set shuffle=False because the data is time-series and shuffline makes if disordered

```python
#split the dataset into train and test
from sklearn.model_selection import train_test_split
# we set shuffle False because the data is time-series
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,shuffle=False,random_state=42)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

# Training and Prediction

```python
#training
from sklearn.linear_model import LinearRegression
model=LinearRegression(fit_intercept=True)
model.fit(X_train,y_train)
print("Model cofficients:",model.coef_)
print("Model intercept:",model.intercept_)

y_pred=model.predict(X_test)
print(y_pred[:5])
print(y_test[:5])
```

# Model Cofficients and Intercept
The linear regression model predicts the target as a weighted sum of the input features plus a small intercept. Feature1 has a negative influence, meaning that as Feature1 increases, the target decreases. Feature2 and Feature3 positively contribute to the target, while Feature4 has negligible effect. The intercept is almost zero, indicating that when all features are zero, the predicted target is approximately zero.

```python
#Evaluation of the model using R-Squared and Mean Squared Error(MSE)
from sklearn.metrics import mean_squared_error,r2_score
r2=r2_score(y_test,y_pred)
mse=mean_squared_error(y_test, y_pred)
print("MSE:",mse)
print("R2:",r2)
```

# Model Evaluation

The model was evaluated using Mean Squared Error (MSE) and R² score:

- MSE = 0.508...: The average squared difference between predicted and actual closing prices is very small, indicating accurate predictions.
- **R² = 0.99997...: The model explains almost all the variance in the target variable, suggesting an excellent fit.

Together, these metrics indicate that the model predicts closing prices very accurately. However, the extremely high R² suggests careful attention to overfitting in future predictions.
